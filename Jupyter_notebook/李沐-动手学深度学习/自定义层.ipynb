```python
# 不带参数的层
```

```python
import torch
import torch.nn.functional as F
from torch import nn
```

```python
# 构造一个没有任何参数的自定义层
class CenteredLayer(nn.Module):
    def __init__(self):
        super().__init__()
    def forward(self,X):
        return X-X.mean()
```

```python
layer=CenteredLayer()
layer(torch.FloatTensor([1,2,3,4,5])) #会调用pytorch的call函数，然后调用本层的forward函数
```

```python
net = nn.Sequential(nn.Linear(8, 128), CenteredLayer())
Y=net(torch.rand(4,8)) #传入的（4个批次，8个特征）
                        #nn.Linear(in_features, out_features) 的输入必须是(batch_size, in_features)
Y.mean()
```

```python
## 带参数的层
```

```python
class MyLinear(nn.Module):
    def __init__(self, in_units, units):
        super().__init__()
        self.weight = nn.Parameter(torch.randn(in_units, units))
        self.bias = nn.Parameter(torch.randn(units,))
    def forward(self, X):
        linear = torch.matmul(X, self.weight.data) + self.bias.data
        return F.relu(linear)

linear = MyLinear(5, 3)
print(linear.weight)
linear.bias
```

```python
# 自定义层直接执行前向传播计算
linear(torch.rand(2, 5))
```

```python
# 使用自定义层构建模型，就像使用内置的全连接层一样使用自定义层
net = nn.Sequential(MyLinear(64, 8), MyLinear(8, 1))
net(torch.rand(2, 64))
```
